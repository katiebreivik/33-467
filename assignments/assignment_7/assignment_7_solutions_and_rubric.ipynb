## problem #1c:

#### observational biases (3 pts each; 6 pts total for 2 biases)
1) The transit method and radial velocity methods both are likely to observe transits at shorter periods, hence the peak in the distribution below 100 days for all the cases

2) Finding multiplanet systems usually reqiures followup observations (like through transit timing variations) that are expensive to obtain, so you are more likely to find more single planet systems based on the current survey strategy

3) Very fast rotating stars have starspots that transit on day-ish timescales, so it is difficult to confirm planets at those short periods

#### exoplanet dynamics (3 pts for 1 reason)
1) Secular resonances occur at fixed orbital period ratios and thus as more planets are added to the system, more short period systems (at higher period ratios) are required

2) Very wide orbits of exoplanets can be disrupted due to fly by's of stellar objects
