```python
import astropy.units as u
import astropy.constants as c
```

The strain is 

$h^2(\mathcal{M}_c, f_{\rm GW}, D) = \frac{128}{5} \frac{(G \mathcal{M}_c)^{10/3}}{c^8 D^2} \left(\pi f_{\rm GW} \right)^{4/3}$

and the chirp is

$\dot{f}(\mathcal{M}_c, f_{\rm GW}) = \frac{96}{5 \pi} \frac{\left(G \mathcal{M}_c \right)^{5/3}}{c^5} (2 \pi f_{\rm orb})^{11/3}$

where $f_{\rm GW} = 2\,f_{\rm orb}$. 


We can solve for the chirp mass from $\dot{f}$ as

$\mathcal{M}_c = \big( \frac{96}{5} \frac{\dot{f}}{(2 \pi f_{\rm orb})^{11/3}} \big)^{3/5} \frac{c^3}{G}$

and for the distance as

$D = \sqrt{\frac{128}{5} \frac{(G \mathcal{M}_c)^{10/3}}{c^8 h^2} \left(\pi f_{\rm GW} \right)^{4/3}}$



```python
def chirp_mass_from_chirp(f_dot, f_orb):
    return ((96/5) * f_dot/(2 * np.pi * f_orb)**(11/3))**(3/5) * c.c**3 / c.G

def distance_from_strain_chirp_mass(strain, chirp_mass, f_gw):
    dist = np.sqrt(128/5 / strain**2 * (np.pi * f_gw)**(4/3) * (c.G * chirp_mass)**(10/3)/c.c**8 )
    return dist
```

```python
fdot = 1.1e-16 * (u.s)**(-2)
f_gw = 2.75 * u.mHz
f_orb = f_gw / 2
strain = 1.8e-22
```

```python
chirp_mass = chirp_mass_from_chirp(fdot, f_orb)
print(chirp_mass.to(u.Msun))
```

```python
distance = distance_from_strain_chirp_mass(strain, chirp_mass, f_gw)
print(distance.to(u.kpc))
```

Based on the chirp and the observed frequency, the chirp mass suggests that the source is a binary black hole!
